```python
!pip install mne
import numpy as np
import mne
import matplotlib.pyplot as plt

# 取樣頻率：1000 Hz
sfreq = 1000

# 產生 10 秒時間
times = np.arange(0, 10, 1/sfreq)

# 模擬 10 Hz 的 alpha wave
signal = np.sin(2 * np.pi * 10 * times)

# 建立 EEG 資料
data = signal[np.newaxis, :]

# 建立通道資訊
info = mne.create_info(
    ch_names=["EEG 001"],
    sfreq=sfreq,
    ch_types=["eeg"]
)

# 建立 Raw EEG
raw = mne.io.RawArray(data, info)

# 畫出 EEG
plt.figure(figsize=(12, 4))
plt.plot(times, signal)

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Simulated 10 Hz Alpha Wave")

plt.show()
```
